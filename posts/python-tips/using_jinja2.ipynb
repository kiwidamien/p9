# Jimja2

The jinja2 templating system is a flexible way of porting information used by many other systems. 

There are a few steps to use Jinja2 to inject variables into a string. While they are not hard, I frequently forget the syntax, so taking note of it here.

## Simple use case: inject into a string

```python
import jinja2

template_string = "Hello, {{name}}!"

environment = jinja2.Environment()
template = environment.from_string(template_string)
template.render(name="World")
```

## Using with a file

Of course, we could always just load the string from the filesystem, and then inject as above. But there is some tooling in Jinja as well

```python
environment = jinja2.Environment(loader=jinja2.FileSystemLoader("jinja_examples/"))
template = environment.get_template("example.sql")
print(template.render(animal='cat'))
```
